Decorators

Decorators Modify Function 


```python


def greet(fx):
    def mfx(*args,**kwargs):
        print("Good Morning")
        fx(*args,**kwargs)
        print("thanks for using this function")

    return mfx

@greet
def Hello():
    print("Hello World")

@greet
def sum(a,b):
    print(a+b)

Hello()

print()

# greet(Hello)()

# greet(sum)(5,2)

print()

sum(4,5)

```

Getters ans Setters

```python
class MyClass:
    def __init__(self,value):
        self._value=value
    def show(self):
        print(f"Value is {self._value}")

    @property
    def ten_value(self):
        return 10*self._value
    
obj=MyClass(10)
print(obj._value)
print(obj.ten_value)
obj.show()
        

```
